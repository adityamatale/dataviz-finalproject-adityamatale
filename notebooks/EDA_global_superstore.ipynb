### Cell 1 — Setup & Load

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

pd.set_option('display.max_columns', None)

df = pd.read_csv('../global_superstore.csv')
print(f"Loaded: {df.shape[0]:,} rows, {df.shape[1]} columns")
df.head()
```

### Cell 2 — Structure & Data Types

```python
df.info()
```

### Cell 3 — Nulls & Duplicates Check

```python
print("Missing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0] if df.isnull().sum().sum() > 0 else "None — dataset is clean")

print(f"\nDuplicate rows: {df.duplicated().sum()}")
```

### Cell 4 — Numerical Summary

```python
df[['Sales', 'Profit', 'Discount', 'Quantity', 'Shipping Cost']].describe().round(2)
```

### Cell 5 — Categorical Attribute Overview

```python
categorical_cols = ['Category', 'Sub-Category', 'Segment', 'Region', 'Market',
                     'Market2', 'Ship Mode', 'Order Priority']

for col in categorical_cols:
    print(f"\n{col} ({df[col].nunique()} unique):")
    print(df[col].value_counts().head(6))
```

### Cell 6 — Spatial Coverage

```python
print(f"Countries: {df['Country'].nunique()}")
print(f"States: {df['State'].nunique()}")
print(f"Cities: {df['City'].nunique()}")
print(f"\nMarket distribution:\n{df['Market'].value_counts()}")
```

### Cell 7 — Temporal Fields: Investigating Order Date / Ship Date

```python
# Order Date and Ship Date arrived corrupted in this export (text-only, no real dates)
print("Order Date unique values:", df['Order Date'].unique())
print("Ship Date unique values:", df['Ship Date'].unique())

# Year and weeknum are intact and will substitute for time-series analysis
print("\nYear range:", df['Year'].min(), "-", df['Year'].max())
print("Year distribution:\n", df['Year'].value_counts().sort_index())
print("\nweeknum range:", df['weeknum'].min(), "-", df['weeknum'].max())
```

### Cell 8 — Cleaning

```python
# ji_lu-shu is a constant column (always 1) — no analytical value
print("ji_lu-shu unique values:", df['ji_lu-shu'].unique())
df = df.drop(columns=['ji_lu-shu'])

# Build a usable pseudo-date from Year + weeknum since Order Date/Ship Date are unusable
df['Year_Week'] = df['Year'].astype(str) + '-W' + df['weeknum'].astype(str).str.zfill(2)
df['Period'] = pd.to_datetime(df['Year'].astype(str) + df['weeknum'].astype(str).str.zfill(2) + '1', format='%Y%W%w')

print(f"\nCleaned shape: {df.shape}")
df[['Year', 'weeknum', 'Year_Week', 'Period']].head()
```

### Cell 9 — Save Cleaned Dataset

```python
df.to_csv('global_superstore_cleaned.csv', index=False)
print(f"Saved cleaned dataset: {df.shape[0]:,} rows, {df.shape[1]} columns")
```
